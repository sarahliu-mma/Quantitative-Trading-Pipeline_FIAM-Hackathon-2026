```python
feat_col = [c for c in jkp['variable'] if c in panel_dt.columns]
```

```python
#rows with missing value ratio>0.4
row_na = panel_dt[feat_col].isna().mean(axis = 1)
mask = row_na >0.4
```

```python
panel_dt[mask]
```

```python
#universe condition
c0 = panel_dt['common'] == 1
c1 = panel_dt['primary_sec'] == 1
c2 = panel_dt['exch_main'] == 1
c3 = (~panel_dt['me'].isna())
```

```python
c4 = panel_dt['prc']>= 5
c5 = panel_dt['bidask'] == 0
c6 = panel_dt['size_grp'] != 'nano'
```

```python
universe_panel = panel_dt.copy()
universe_panel = universe_panel[c0 & c1 & c2 & c3 & c4 & c5 & c6]
```
